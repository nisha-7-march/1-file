```python
import numpy as np
from sklearn.impute import SimpleImputer

# 1. Create a 3x3 matrix with missing values (Added missing commas)
matrix = np.array([
    [1, 2, np.nan],
    [4, np.nan, 6],
    [np.nan, 8, 9]
])

print("--- Original 3x3 Matrix ---")
print(matrix)

# 2. Initialize the imputer (fills NaN with the average of that column)
imputer = SimpleImputer(missing_values=np.nan, strategy='mean')

# 3. Calculate means and replace the missing values
imputed_matrix = imputer.fit_transform(matrix)

print("\n--- Imputed 3x3 Matrix (Column Mean) ---")
print(imputed_matrix)

```
